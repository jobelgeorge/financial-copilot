```python
def prepare_metric(df, metric_name):
    """Prepare a metric DataFrame for merging."""

    result = df[
        ["fiscal_year", "value_billions"]
    ].copy()

    result = result.rename(
        columns={
            "value_billions": metric_name
        }
    )

    return result
```

```python
revenue_clean = prepare_metric(
    revenue,
    "revenue"
)

net_income_clean = prepare_metric(
    net_income,
    "net_income"
)

assets_clean = prepare_metric(
    assets,
    "assets"
)

liabilities_clean = prepare_metric(
    liabilities,
    "liabilities"
)

cash_clean = prepare_metric(
    cash,
    "cash"
)

operating_cash_flow_clean = prepare_metric(
    operating_cash_flow,
    "operating_cash_flow"
)
```

```python
financials = revenue_clean
```

```python
financials = financials.merge(
    net_income_clean,
    on="fiscal_year",
    how="outer"
)

financials = financials.merge(
    assets_clean,
    on="fiscal_year",
    how="outer"
)

financials = financials.merge(
    liabilities_clean,
    on="fiscal_year",
    how="outer"
)

financials = financials.merge(
    cash_clean,
    on="fiscal_year",
    how="outer"
)

financials = financials.merge(
    operating_cash_flow_clean,
    on="fiscal_year",
    how="outer"
)
```
